## StuffDocumentChain Text summarization

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("apjspeech.pdf")
docs = loader.load()
docs
```

```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_classic.chains.combine_documents import (
    create_stuff_documents_chain,
)

# Create the summarization prompt
prompt = ChatPromptTemplate.from_template(
    """
    Write a concise summary of the following speech.
    Include the main ideas and important points.

    Speech:
    {context}
    """
)

# Create the Stuff documents chain
chain = create_stuff_documents_chain(
    llm=llm,
    prompt=prompt
)

# Generate the summary
result = chain.invoke({"context": docs})

print(result)
```

### Map reduce to Summarize Large documents

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
docs
```

```python
len(docs)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=2000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(docs)
chunks
```

```python
len(chunks)
```

```python
# --------------------------------------------------
# Step 2: Create the MAP prompt
# --------------------------------------------------

map_prompt = ChatPromptTemplate.from_template(
    """
    Summarize the following speech excerpt concisely.
    Preserve the important ideas and facts.

    Speech excerpt:
    {text}

    Summary:
    """
)
```

```python
# --------------------------------------------------
# Step 3: Build the MAP chain using LCEL
# --------------------------------------------------
from langchain_core.output_parsers import StrOutputParser

map_chain = map_prompt | llm | StrOutputParser()
```

```python
# --------------------------------------------------
# Step 4: Run the MAP stage on every chunk
# --------------------------------------------------

map_inputs = [
    {"text": chunk.page_content}
    for chunk in chunks
]

map_summaries = map_chain.batch(
    map_inputs,
    config={"max_concurrency": 1}
)

print("Intermediate summaries:", len(map_summaries))
```

```python
# --------------------------------------------------
# Step 5: Create the REDUCE prompt
# --------------------------------------------------

reduce_prompt = ChatPromptTemplate.from_template(
    """
    You are given summaries of different parts of a speech.

    Combine them into one coherent, concise summary.
    Remove repetition and preserve the key ideas.

    Intermediate summaries:
    {summaries}

    Final summary:
    """
)
```

```python
# --------------------------------------------------
# Step 6: Build the REDUCE chain using LCEL
# --------------------------------------------------

reduce_chain = reduce_prompt | llm | StrOutputParser()
```

```python
# --------------------------------------------------
# Step 7: Combine the intermediate summaries
# --------------------------------------------------

combined_summaries = "\n\n".join(map_summaries)

final_summary = reduce_chain.invoke(
    {"summaries": combined_summaries}
)

print("\nFINAL SUMMARY:\n")
print(final_summary)
```

### Refine Chain For Summarization

```python
# --------------------------------------------------
# Step 1: Create the initial summary prompt
# --------------------------------------------------

initial_prompt = ChatPromptTemplate.from_template(
    """
    Write a concise summary of the following speech excerpt.
    Capture its main ideas and important facts.

    Speech excerpt:
    {text}

    Initial summary:
    """
)
```

```python
# --------------------------------------------------
# Step 2: Create the refine prompt
# --------------------------------------------------

refine_prompt = ChatPromptTemplate.from_template(
    """
    You are refining an existing summary of a speech.

    Existing summary:
    {existing_summary}

    New speech excerpt:
    {text}

    Update the existing summary using relevant information
    from the new excerpt.

    Instructions:
    - Preserve important information from the existing summary.
    - Add new key ideas and facts from the new excerpt.
    - Remove repetition.
    - Keep the summary concise and coherent.

    Refined summary:
    """
)
```

```python
# --------------------------------------------------
# Step 4: Build LCEL chains
# --------------------------------------------------

initial_chain = initial_prompt | llm | StrOutputParser()

refine_chain = refine_prompt | llm | StrOutputParser()
```

```python
# --------------------------------------------------
# Step 5: Generate the initial summary
# --------------------------------------------------

summary = initial_chain.invoke(
    {"text": chunks[0].page_content}
)

summary
```

```python
# --------------------------------------------------
# Step 6: Refine the summary with remaining chunks
# --------------------------------------------------

for index, chunk in enumerate(chunks[1:], start=2):

    summary = refine_chain.invoke(
        {
            "existing_summary": summary,
            "text": chunk.page_content
        }
    )
    
    print(f"Processed chunk {index}/{len(chunks)}")

```

```python
# --------------------------------------------------
# Step 7: Display the final summary
# --------------------------------------------------

print("\nFINAL REFINED SUMMARY:\n")
print(summary)
```
